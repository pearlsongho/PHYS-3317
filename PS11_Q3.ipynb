```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

# c) Compute E(kx, ky), plot n(E)

```python
#define variables

t = 1
e_0 = 0
a = 1
N = 200
```

```python
#calculate E(k_x, k_y)

def E(k_x, k_y):
  return e_0 - 2*t*np.cos(k_x*a) - 2*t*np.cos(k_y*a)

energies = []

for i in range(N):
  for j in range(N):
    k_x = 2*np.pi*i/N
    k_y = 2*np.pi*j/N
    energies.append(E(k_x, k_y))
```

```python
#make plot

plt.hist(energies, bins = 100, color = 'purple', edgecolor = 'black')
plt.xlabel('Energy E')
plt.ylabel('Number of States n(E)')
plt.title('n(E) for 2D Square Lattice')
```

# d) Plot Fermi surface (1 electron per Cu)

```python
#find Fermi energy E_F

energies.sort()
E_F = energies[int(N**2/2)]
print("Fermi energy: " + str(E_F))
```

```python
#plot Fermi surface

x = np.linspace(-np.pi/a, np.pi/a, 100)
y = np.linspace(-np.pi/a, np.pi/a, 100)

k_x, k_y = np.meshgrid(x, y)

E = e_0 - 2*t*np.cos(k_x*a) - 2*t*np.cos(k_y*a)

plt.contourf(k_x, k_y, E, levels = [E.min(), E_F], cmap = 'plasma')

plt.contour(k_x, k_y, E, levels = [E_F])
plt.xlabel(r'$k_x$')
plt.ylabel(r'$k_y$')
plt.title('Filled States')

plt.plot([], [], color = 'purple', label = r'$E = E_F$')
plt.axis('square')
plt.legend()
```
