```python
import numpy as np
import matplotlib.pyplot as plt

#create grid
x = np.linspace(0, a, 300)
X1, X2 = np.meshgrid(x, x)

#single-particle wavefunctions
psi1_x1 = np.sqrt(2/a) * np.sin(((np.pi)/a) * X1)
psi1_x2 = np.sqrt(2/a) * np.sin(((np.pi)/a) * X2)

psi2_x1 = np.sqrt(2/a) * np.sin((2 * (np.pi)/a) * X1)
psi2_x2 = np.sqrt(2/a) * np.sin((2 * (np.pi)/a) * X2)

#construct total wavefunctions
Psi_D = psi1_x1 * psi2_x2
Psi_B = (1/np.sqrt(2)) * (psi1_x1 * psi2_x2 + psi2_x1 * psi1_x2)
Psi_F = (1/np.sqrt(2)) * (psi1_x1 * psi2_x2 - psi2_x1 * psi1_x2)
```

```python
#distinguishable particles

vmax = np.max(np.abs(Psi_D))
plt.figure(figsize=(6, 5))
plt.imshow(Psi_D, origin="lower", extent=[0, a, 0, a],
           cmap="PiYG", vmin=-vmax, vmax=vmax)
plt.colorbar(label="value of $\\psi_{D}$")
plt.plot([0, a], [0, a], "k--")
plt.xlabel("$x_1 / a$")
plt.ylabel("$x_2 / a$")
plt.title("$\\psi_{D}(x_1, x_2) = \\psi_1(x_1)\\,\\psi_2(x_2)$")
plt.show()
```

```python
#indistinguishable bosons

vmax = np.max(np.abs(Psi_B))
plt.figure(figsize=(6, 5))
plt.imshow(Psi_B, origin="lower", extent=[0, a, 0, a],
           cmap="Spectral", vmin=-vmax, vmax=vmax)
plt.colorbar(label="value of $\\psi_{B}$")
plt.plot([0, a], [0, a], "k--")      # dashed black line x1 = x2
plt.xlabel("$x_1 / a$")
plt.ylabel("$x_2 / a$")
plt.title("$\\psi_B(x_1,x_2) = \\frac{1}{\\sqrt{2}}[\\psi_1(x_1)\\psi_2(x_2) + \\psi_2(x_1)\\psi_1(x_2)]$")
plt.show()
```

```python
#indistinguishable fermions

vmax = np.max(np.abs(Psi_F))
plt.figure(figsize=(6, 5))
plt.imshow(Psi_F, origin="lower", extent=[0, a, 0, a],
           cmap="plasma", vmin=-vmax, vmax=vmax)
plt.colorbar(label="value of $\\psi_{F}$")
plt.plot([0, a], [0, a], "k--")      # dashed black line x1 = x2
plt.xlabel("$x_1 / a$")
plt.ylabel("$x_2 / a$")
plt.title("$\\psi_F(x_1,x_2) = \\frac{1}{\\sqrt{2}}[\\psi_1(x_1)\\psi_2(x_2) - \\psi_2(x_1)\\psi_1(x_2)]$")
plt.show()
```
